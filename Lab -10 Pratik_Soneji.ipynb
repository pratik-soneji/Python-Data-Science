### 2. Creating Series from Different Data

Create three Pandas Series: i) from a Python list, ii) from a NumPy array, and iii) from a dictionary. Display all three Series.

```python
list_data = [10, 20, 30, 40]
s_list = pd.Series(list_data)


array_data = np.array([1.5, 2.5, 3.5, 4.5])
s_array = pd.Series(array_data)


dict_data = {"Math": 88, "Science": 92, "English": 79}
s_dict = pd.Series(dict_data)

print("Series from List")
print(s_list)
print("\n Series from NumPy Array ")
print(s_array)

print("\n Series from Dictionary ")
print(s_dict)
```

### 4. Creating Time Series

Create a date range for 10 consecutive days starting from '2026-09-01' and create a Pandas Series containing daily sales values for those dates. Display the resulting time series.

```python
dates = pd.date_range(start="2026-09-01", periods=10, freq="D")

sales_values = [150, 200, 180, 220, 210, 250, 300, 290, 310, 280]

sales_series = pd.Series(data=sales_values, index=dates)
print(sales_series)
```

### 5. Time Series with Frequency

Create a time series for the first 6 months of 2026 using a suitable date frequency. Store monthly revenue values in the Series and display the result.

```python
months = pd.date_range(start="2026-01-01", periods=6, freq="MS")

revenue_values = [12000, 15000, 14000, 18000, 21000, 19500]
revenue_series = pd.Series(data=revenue_values, index=months)
print(revenue_series)
```

### 8. Conditional Selection

Using the `students` DataFrame, display: i) students having Marks greater than 70, ii) students belonging to the 'CSE' department, and iii) students having Marks between 60 and 80.

```python

students = pd.DataFrame(data)

marks_gt_70 = students[students["Marks"] > 70]

cse_students = students[students["Department"] == "CSE"]

marks_60_80 = students[(students["Marks"] >= 60) & (students["Marks"] <= 80)]
print(" Marks > 70")
print(marks_gt_70)
print("\n CSE Department ")
print(cse_students)
print("\n Marks Between 60 and 80")
print(marks_60_80)
```

### 9. Multiple Conditions

Using the `students` DataFrame, select students who belong to the 'CSE' department and have Marks greater than or equal to 70. Also display students who belong to either 'CSE' or 'IT'.

```python

students = pd.DataFrame(data)


cse_and_high_marks = students[
    (students["Department"] == "CSE") & (students["Marks"] >= 70)
]

cse_or_it = students[
    (students["Department"] == "CSE") | (students["Department"] == "IT")
]

print("CSE department with Marks >= 70 ")
print(cse_and_high_marks)

print("\n belong to either CSE or IT ")
print(cse_or_it)
```

### 10. Conditional Selection on DataFrame Columns

From the `products` DataFrame, display products whose Price is greater than 500 and Quantity is less than 20. Then display only the ProductName and Price columns of the selected records.

```python

products_df = pd.DataFrame(product_data)

condition = (products_df["Price"] > 500) & (products_df["Quantity"] < 20)
filter_products = products_df[condition]

result = filter_products[["ProductName", "Price"]]

print("Selected Products (Price > 500 & Quantity < 20) ")
print(result)
```
