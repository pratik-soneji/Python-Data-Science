### 04) WAP to plot a line chart showing temperature variation during a week. Mark the lowest and highest temperatue.

```python
Days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
Temperature = [20, 32, 31, 33, 34, 43, 36]
```

```python
plt.plot(Days, Temperature, marker="o")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Variation During a Week")

min_temp = min(Temperature)
max_temp = max(Temperature)
min_index = Temperature.index(min_temp)
max_index = Temperature.index(max_temp)

plt.scatter(Days[min_index], min_temp, s=100, label=f"Lowest: {min_temp}°C")
plt.scatter(Days[max_index], max_temp, s=100, label=f"Highest: {max_temp}°C")

plt.legend()
plt.grid(True)
plt.show()
```

### 05) WAP to create a detailed pie chart representing the monthly expenditure distribution of a family. 
### The chart should display labels, percentage values, title, different colors for each slice, and explode one slice to highlight the highest expense.

```python
Categories = ["Rent", "Food", "Education", "Transport", "Entertainment", "Savings"]
Amount = [15000, 8000, 6000, 3000, 2000, 5000]
```

```python
highest_index = Amount.index(max(Amount))
explode = [0.1 if i == highest_index else 0 for i in range(len(Amount))]

plt.pie(
    Amount,
    labels=Categories,
    autopct="%1.1f%%",
    explode=explode,
    colors=["lightblue", "lightgreen", "gold", "orange", "violet", "pink"],
    startangle=90
)
plt.title("Monthly Expenditure Distribution of a Family")
plt.axis("equal")
plt.show()
```

### 06) WAP to create bar chart showing the marks obtained by students in different subjects.
### The program should also display the marks on the top of the respective bar.

```python
Subjects = ["Python", "DBMS", "SE", "PS", "PE-1"]
Marks = [78, 85, 72, 80, 90]
```

```python
bars = plt.bar(Subjects, Marks)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks Obtained in Different Subjects")
plt.ylim(0, 100)

for bar, mark in zip(bars, Marks):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 1,
        str(mark),
        ha="center"
    )

plt.show()
```

### 07) WAP to create a horizontal bar chart representing the number of books sold in different categories. 
### The program should also display the number of books sold at the end of the respective bar.

```python
Categories = ["Fiction", "Science", "History", "Technology", "Biography"]
Books_Sold = [120, 90, 70, 110, 60]
```

```python
bars = plt.barh(Categories, Books_Sold)

plt.xlabel("Number of Books Sold")
plt.ylabel("Categories")
plt.title("Books Sold in Different Categories")

for bar, value in zip(bars, Books_Sold):
    plt.text(
        bar.get_width() + 2,
        bar.get_y() + bar.get_height() / 2,
        str(value),
        va="center"
    )

plt.show()
```
